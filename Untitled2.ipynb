```python
# Install KaggleHub if needed
!pip -q install kagglehub

import os
import re
import math
import kagglehub
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from collections import Counter
from urllib.parse import urlparse

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    ConfusionMatrixDisplay
)

# -----------------------------------------------------
# 1. Download PhiUSIIL dataset from Kaggle
# -----------------------------------------------------

path = kagglehub.dataset_download(
    "kaggleprollc/phishing-url-websites-dataset-phiusiil"
)

print("Dataset location:", path)
print(os.listdir(path))

csv_file = os.path.join(path, "PhiUSIIL_Phishing_URL_Dataset.csv")
df = pd.read_csv(csv_file)

print("Dataset shape:", df.shape)
display(df.head())

print("\nOriginal label distribution:")
print(df["label"].value_counts())

# Official dataset:
# 1 = legitimate
# 0 = phishing
# Convert to:
# 1 = phishing
# 0 = legitimate

df["phishing"] = (df["label"] == 0).astype(int)

print("\nConverted target distribution:")
print(df["phishing"].value_counts())


# -----------------------------------------------------
# 2. FEATURE ENGINEERING
# -----------------------------------------------------

def calculate_entropy(url):
    """Calculate Shannon entropy of the characters in a URL."""
    if pd.isna(url) or len(url) == 0:
        return 0

    counts = Counter(str(url))
    length = len(str(url))

    probabilities = [count / length for count in counts.values()]

    return -sum(
        p * math.log2(p)
        for p in probabilities
        if p > 0
    )


def calculate_path_depth(url):
    """Count directory/path components appearing after the domain."""
    if pd.isna(url):
        return 0

    url = str(url)

    if "://" not in url:
        url = "http://" + url

    try:
        path = urlparse(url).path
        return len([part for part in path.split("/") if part])
    except:
        return 0


suspicious_pattern = (
    r"login|verify|verification|secure|account|update|"
    r"password|signin|confirm|bank|authentication|credential"
)

df["URL_Entropy"] = df["URL"].apply(calculate_entropy)

df["PathDepth"] = df["URL"].apply(calculate_path_depth)

df["SuspiciousKeywordCount"] = (
    df["URL"]
    .fillna("")
    .str.lower()
    .str.count(suspicious_pattern)
)

print("\nEngineered features:")
display(
    df[
        [
            "URL",
            "URL_Entropy",
            "PathDepth",
            "SuspiciousKeywordCount",
            "phishing"
        ]
    ].head(10)
)


# -----------------------------------------------------
# 3. BASELINE FEATURES
# -----------------------------------------------------

baseline_features = [
    "URLLength",
    "DomainLength",
    "IsDomainIP",
    "NoOfSubDomain",
    "HasObfuscation",
    "NoOfObfuscatedChar",
    "NoOfLettersInURL",
    "NoOfDegitsInURL",
    "NoOfQMarkInURL",
    "NoOfEqualsInURL",
    "NoOfAmpersandInURL",
    "NoOfOtherSpecialCharsInURL",
    "IsHTTPS"
]

engineered_features = baseline_features + [
    "URL_Entropy",
    "PathDepth",
    "SuspiciousKeywordCount"
]

X_baseline = df[baseline_features].fillna(0)
X_engineered = df[engineered_features].fillna(0)

y = df["phishing"]


# -----------------------------------------------------
# 4. TRAIN/TEST SPLIT
# -----------------------------------------------------

train_index, test_index = train_test_split(
    np.arange(len(df)),
    test_size=0.20,
    random_state=42,
    stratify=y
)

Xb_train = X_baseline.iloc[train_index]
Xb_test = X_baseline.iloc[test_index]

Xe_train = X_engineered.iloc[train_index]
Xe_test = X_engineered.iloc[test_index]

y_train = y.iloc[train_index]
y_test = y.iloc[test_index]


# -----------------------------------------------------
# 5. TRAIN BASELINE RANDOM FOREST
# -----------------------------------------------------

baseline_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

baseline_model.fit(Xb_train, y_train)

baseline_predictions = baseline_model.predict(Xb_test)


# -----------------------------------------------------
# 6. TRAIN FEATURE-ENGINEERED RANDOM FOREST
# -----------------------------------------------------

engineered_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

engineered_model.fit(Xe_train, y_train)

engineered_predictions = engineered_model.predict(Xe_test)


# -----------------------------------------------------
# 7. COMPARE PERFORMANCE
# -----------------------------------------------------

def metrics(y_true, predictions):
    return {
        "Accuracy": accuracy_score(y_true, predictions),
        "Precision": precision_score(y_true, predictions),
        "Recall": recall_score(y_true, predictions),
        "F1 Score": f1_score(y_true, predictions)
    }

results = pd.DataFrame(
    {
        "Baseline Random Forest":
            metrics(y_test, baseline_predictions),

        "Feature-Engineered Random Forest":
            metrics(y_test, engineered_predictions)
    }
).T

print("\nMODEL COMPARISON")
display(results.round(4))


# -----------------------------------------------------
# 8. CLASSIFICATION REPORT
# -----------------------------------------------------

print("\nFEATURE-ENGINEERED MODEL CLASSIFICATION REPORT")
print(
    classification_report(
        y_test,
        engineered_predictions,
        target_names=["Legitimate", "Phishing"]
    )
)


# -----------------------------------------------------
# 9. CONFUSION MATRIX
# -----------------------------------------------------

ConfusionMatrixDisplay.from_predictions(
    y_test,
    engineered_predictions,
    display_labels=["Legitimate", "Phishing"]
)

plt.title("Feature-Engineered Random Forest Confusion Matrix")
plt.show()


# -----------------------------------------------------
# 10. FEATURE IMPORTANCE
# -----------------------------------------------------

importance = pd.DataFrame(
    {
        "Feature": engineered_features,
        "Importance": engineered_model.feature_importances_
    }
).sort_values("Importance", ascending=False)

display(importance)

plt.figure(figsize=(10, 7))

top_features = importance.head(15).sort_values("Importance")

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.xlabel("Feature Importance")
plt.ylabel("Feature")
plt.title("Top Random Forest Features")
plt.tight_layout()
plt.show()
```
